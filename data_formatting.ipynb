```python
import csv
```

```python
input_file = "performance_difference_plot_data.csv"
g1_file = "performance_difference_plot_data_G1_IK.csv"
g2_file = "performance_difference_plot_data_G2_IK.csv"

rows_out = []

with open(input_file, newline="") as f:
    reader = csv.DictReader(f)

    index = 1
    for row in reader:
        # remove leader/self-comparison rows
        if row["member"] == row["leader"]:
            continue

        new_row = {
            "index": index,
            "member_error_diff": row["member_error_diff"],
            "bound": row["bound"],
            "group_id": row["group_id"],
        }
        rows_out.append(new_row)
        index += 1

fieldnames = ["index", "member_error_diff", "bound"]

with open(g1_file, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    for row in rows_out:
        if row["group_id"] == "0":
            writer.writerow({k: row[k] for k in fieldnames})

with open(g2_file, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    for row in rows_out:
        if row["group_id"] == "1":
            writer.writerow({k: row[k] for k in fieldnames})
```

```python
input_file = "group_violin_data.csv"

splits = [
    ("group_violin_data_A.csv", "groups",  "0"),
    ("group_violin_data_B.csv", "groups",  "1"),
    ("group_violin_data_C.csv", "groupsm", "0"),
    ("group_violin_data_D.csv", "groupsm", "1"),
    ("group_violin_data_E.csv", "groupsm", "2"),
    ("group_violin_data_F.csv", "groupsm", "3"),
    ("group_violin_data_G.csv", "groupsm", "4"),
]

with open(input_file, newline="") as f:
    rows = list(csv.DictReader(f))

for output_file, method, group_id in splits:
    with open(output_file, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=["value"])
        writer.writeheader()

        for row in rows:
            if row["method"] == method and row["group_id"] == group_id:
                writer.writerow({
                    "value": row["distance_to_leader"]
                })
```

```python
input_file = "performance_degradation_violin.csv"

splits = [
    ("group_violin_data_performance_A.csv", "groups",  "0"),
    ("group_violin_data_performance_B.csv", "groups",  "1"),
    ("group_violin_data_performance_C.csv", "groupsm", "0"),
    ("group_violin_data_performance_D.csv", "groupsm", "1"),
    ("group_violin_data_performance_E.csv", "groupsm", "2"),
    ("group_violin_data_performance_F.csv", "groupsm", "3"),
    ("group_violin_data_performance_G.csv", "groupsm", "4"),
]

with open(input_file, newline="") as f:
    rows = list(csv.DictReader(f))

for output_file, method, group_id in splits:
    with open(output_file, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=["value"])
        writer.writeheader()

        for row in rows:
            if row["method"] == method and row["group_id"] == group_id:
                writer.writerow({
                    "value": format(float(row["delta_error"]), ".15g")
                })
```
